```python
#====================================================================================
# VIDA CRM Lakehouse
# Bronze Layer - Data Ingestion
# Author: Emmanuel Quesada Gómez
#====================================================================================

# Define volume path
VOLUME_PATH = "/Volumes/workspace/default/raw_data"

# Read all CSVs from volume
df_volunteer = spark.read.csv(f"{VOLUME_PATH}/dim_volunteer.csv", header=True, inferSchema=True)
df_geography = spark.read.csv(f"{VOLUME_PATH}/dim_geography.csv", header=True, inferSchema=True)
df_organization = spark.read.csv(f"{VOLUME_PATH}/dim_organization.csv", header=True, inferSchema=True)
df_project = spark.read.csv(f"{VOLUME_PATH}/dim_project.csv", header=True, inferSchema=True)
df_date = spark.read.csv(f"{VOLUME_PATH}/dim_date.csv", header=True, inferSchema=True)
df_fact = spark.read.csv(f"{VOLUME_PATH}/fact_volunteer_activity.csv", header=True, inferSchema=True)

print("All CSVs loaded successfully")
```

```python
#====================================================================================
# Write to Bronze Delta Tables
#====================================================================================

BRONZE_PATH = "workspace.default"

df_volunteer.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE_PATH}.bronze_volunteer")
df_geography.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE_PATH}.bronze_geography")
df_organization.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE_PATH}.bronze_organization")
df_project.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE_PATH}.bronze_project")
df_date.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE_PATH}.bronze_date")
df_fact.write.format("delta").mode("overwrite").saveAsTable(f"{BRONZE_PATH}.bronze_fact_activity")

print("All Bronze Delta Tables created successfully")
```
